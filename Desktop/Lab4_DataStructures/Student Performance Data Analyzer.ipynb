```python
LASTNAME = "Baruelo"
STUDENTID = "TUPM-26-1254"
SEED_NUM = 4
FAVORITE_ARTIST = "zild"

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

students = ["Student 1", "Student 2", "Student 3",
            "Student 4", "Student 5"]

print("=== STUDENT INFORMATION ===")
print("Last Name:", LASTNAME)
print("Student ID:", STUDENTID)
print("Subjects:", subjects)
```

```python
scores = [
    [85, 90, 88, 92, 87],
    [78, 82, 80, 85, 79],
    [95, 93, 96, 91, 94],
    [70, 75, 72, 78, 74],
    [88, 86, 90, 89, 85]
]

records = []

for i in range(5):
    for j in range(5):
        record = {
            "student": students[i],
            "subject": subjects[j],
            "score": scores[i][j]
        }
        records.append(record)

# Add a repeated record
records.append(records[0].copy())

# Add an incomplete record
records.append({
    "student": "Student 5",
    "subject": "MATHEMATICS",
    "score": None
})

print("Total Records:", len(records))
for record in records:
    print(record)
```

```python
print("=== REPEATED RECORDS ===")

for i in range(len(records)):
    for j in range(i + 1, len(records)):
        if records[i] == records[j]:
            print("Repeated:", records[i])

print("\n=== INCOMPLETE RECORDS ===")

for record in records:
    if record["score"] is None:
        print(record)
```

```python
print("=== PERFORMANCE CATEGORIES ===")

for record in records:
    score = record["score"]

    if score is None:
        category = "INCOMPLETE"
    elif score >= 90:
        category = "EXCELLENT"
    elif score >= 80:
        category = "GOOD"
    elif score >= 75:
        category = "PASSED"
    else:
        category = "FAILED"

    record["category"] = category
    print(record["student"], record["subject"],
          score, category)
```

```python
print("=== STUDENT AVERAGES ===")

for student in students:
    student_scores = []

    for record in records[:-2]:
        if record["student"] == student:
            student_scores.append(record["score"])

    average = sum(student_scores) / len(student_scores)
    print(student, "Average:", round(average, 2))

print("\n=== SUBJECT AVERAGES ===")

for subject in subjects:
    subject_scores = []

    for record in records[:-2]:
        if record["subject"] == subject:
            subject_scores.append(record["score"])

    average = sum(subject_scores) / len(subject_scores)
    print(subject, "Average:", round(average, 2))
```

```python
print("=== FINAL PERFORMANCE REPORT ===")

for student in students:
    total = 0
    count = 0

    for record in records[:-2]:
        if record["student"] == student:
            total += record["score"]
            count += 1

    average = total / count

    if average >= 90:
        result = "EXCELLENT"
    elif average >= 80:
        result = "GOOD"
    elif average >= 75:
        result = "PASSED"
    else:
        result = "FAILED"

    print(student, "Average:", round(average, 2),
          "Result:", result)
```
