```python
products = ["Notebook", "Pen", "Pencil", "Paper", "Folder"]
quantities = [4, 5, 3, 4, 2]
prices = [25, 10, 8, 5, 15]
categories = ["SCHOOL", "SCHOOL", "SCHOOL",
              "OFFICE", "OFFICE"]

transactions = []

for i in range(5):
    record = {
        "product": products[i],
        "quantity": quantities[i],
        "price": prices[i],
        "category": categories[i]
    }

    transactions.append(record)

# Add a repeated transaction
transactions.append(transactions[0].copy())

print("=== TRANSACTION RECORDS ===")
for record in transactions:
    print(record)
```

```python
print("=== PURCHASE TOTALS ===")

for record in transactions:
    total = record["quantity"] * record["price"]
    record["total"] = total

    print("Product:", record["product"])
    print("Quantity:", record["quantity"])
    print("Price:", record["price"])
    print("Total:", total)
    print()
```

```python
print("=== REPEATED TRANSACTIONS ===")

for i in range(len(transactions)):
    for j in range(i + 1, len(transactions)):
        if transactions[i] == transactions[j]:
            print("Repeated:", transactions[i])
```

```python
print("=== CATEGORY SUMMARY ===")

for category in set(categories):
    total = 0

    for record in transactions:
        if record["category"] == category:
            total += record["total"]

    print(category, "Total Sales:", total)
```

```python
print("=== PURCHASE HISTORY ===")

grandtotal = 0

for record in transactions:
    grandtotal += record["total"]

print("Number of Transactions:", len(transactions))
print("Total Sales:", grandtotal)

average = grandtotal / len(transactions)
print("Average Transaction:", round(average, 2))
```

```python
print("=== FINAL TRANSACTION REPORT ===")

for record in transactions:
    if record["total"] >= 100:
        result = "HIGH PURCHASE"
    else:
        result = "REGULAR PURCHASE"

    print(record["product"], "-", result)

print("Grand Total:", grandtotal)
```
